````python
df=df.dropna(subset=['genre','overview'])
````

````python
df.isnull().sum()
````

````python
df=df[['id','title','overview','genre']]
````

````python
df
````

````python
df['tags']=df['overview']+df['genre']
````

````python
df
````

````python
new_data=df.drop(columns=['overview','genre'])
````

````python
new_data
````

````python
from sklearn.feature_extraction.text import CountVectorizer
````

````python
cv=CountVectorizer(max_features=10000,stop_words='english')
````

````python
vector=cv.fit_transform(new_data['tags'].values.astype('U')).toarray()
````

````python
vector.shape
````

now we use cosine similarity algorithm

````python
from sklearn.metrics.pairwise import cosine_similarity
````

````python
similarity=cosine_similarity(vector)
````

````python
similarity
````

````python
new_data[new_data['title']=="The Godfather"].index[0]
````

````python
distance=sorted(list(enumerate(similarity[2])),reverse=True, key=lambda vector:vector[1])
for i in distance[0:10]:
  print(new_data.iloc[i[0]].title)
````

````python
def recommend(movie):

    movie = movie.lower()
    new_data['title'] = new_data['title'].str.lower()

    if movie not in new_data['title'].values:
        print("Movie not found in dataset")
        return

    index = new_data[new_data['title'] == movie].index[0]

    distance = sorted(list(enumerate(similarity[index])), reverse=True, key=lambda x: x[1])

    for i in distance[1:6]:
        print(new_data.iloc[i[0]].title)
````

````python
recommend("Iron man")
````

````python
recommend("The Godfather")
````
